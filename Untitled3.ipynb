```python
import cv2
import numpy as np
import time
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Generate a Sample Video Stream (120 frames, 30 FPS)
# ---------------------------------------------------------
width, height, fps, total_frames = 640, 480, 30, 120
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
raw_video = cv2.VideoWriter('input_stream.mp4', fourcc, fps, (width, height))

# Moving target objects: [x, y, vx, vy, radius, shape_type]
objects = [
    [100, 120, 4, 3, 35, "circle"],
    [450, 200, -3, 4, 40, "rect"],
    [250, 350, 3, -3, 30, "circle"],
    [500, 380, -4, -2, 35, "rect"]
]

for f in range(total_frames):
    # Create a light-gray textured background with slight lighting gradient
    frame = np.full((height, width, 3), 220, dtype=np.uint8)
    for col in range(width):
        frame[:, col] = np.clip(220 - int(col * 0.06), 160, 230)

    # Move and draw dark objects on screen
    for obj in objects:
        obj[0] += obj[2]
        obj[1] += obj[3]
        if obj[0] - obj[4] < 10 or obj[0] + obj[4] > width - 10:
            obj[2] *= -1
        if obj[1] - obj[4] < 10 or obj[1] + obj[4] > height - 10:
            obj[3] *= -1

        x, y, r = int(obj[0]), int(obj[1]), obj[4]
        if obj[5] == "circle":
            cv2.circle(frame, (x, y), r, (35, 35, 35), -1)
        else:
            cv2.rectangle(frame, (x - r, y - r), (x + r, y + r), (25, 25, 25), -1)

    raw_video.write(frame)
raw_video.release()

# ---------------------------------------------------------
# 2. Computer Vision Pipeline (Gaussian + Adaptive Threshold + Contours)
# ---------------------------------------------------------
cap = cv2.VideoCapture('input_stream.mp4')
out_video = cv2.VideoWriter('output_detected.mp4', fourcc, fps, (width, height))

inference_times_ms = []
fps_list = []
detected_counts = []
sample_frames = []

frame_idx = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    start_t = time.perf_counter()

    # Step A: Convert to Grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Step B: Apply Gaussian Blur Filter to remove pixel noise
    blurred = cv2.GaussianBlur(gray, (7, 7), 0)

    # Step C: Apply Adaptive Threshold Matrix to segment dark shapes
    thresh = cv2.adaptiveThreshold(
        blurred, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY_INV,
        blockSize=31, C=8
    )

    # Clean small holes inside segmented shapes
    kernel = np.ones((5, 5), np.uint8)
    thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

    # Step D: Extract target contours based on pixel area bounds
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    circle_count = 0
    rect_count = 0

    for cnt in contours:
        area = cv2.contourArea(cnt)
        # Filter out tiny noise (< 800 px) or huge background artifacts (> 25000 px)
        if 800 < area < 25000:
            x, y, w, h = cv2.boundingRect(cnt)
            peri = cv2.arcLength(cnt, True)
            approx = cv2.approxPolyDP(cnt, 0.04 * peri, True)

            # Classify target class by polygon vertices
            if len(approx) <= 5:
                label = "Square/Rect"
                rect_count += 1
                color = (0, 140, 255)   # Orange box
            else:
                label = "Circle"
                circle_count += 1
                color = (0, 200, 0)     # Green box

            # Draw contour outline, bounding box, and class tag
            cv2.drawContours(frame, [cnt], -1, (255, 0, 255), 2)
            cv2.rectangle(frame, (x, y), (x + w, y + h), color, 2)
            cv2.putText(frame, f"{label} ({int(area)}px)", (x, y - 8),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

    elapsed_ms = (time.perf_counter() - start_t) * 1000
    current_fps = 1000.0 / elapsed_ms if elapsed_ms > 0 else 0

    inference_times_ms.append(elapsed_ms)
    fps_list.append(current_fps)
    total_detected = circle_count + rect_count
    detected_counts.append(total_detected)

    # Step E: Overlay dynamic text count vectors and real-time FPS
    cv2.rectangle(frame, (10, 10), (310, 115), (20, 20, 20), -1)
    cv2.putText(frame, f"Frame: {frame_idx + 1}/{total_frames}", (20, 32),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
    cv2.putText(frame, f"Total Objects: {total_detected}", (20, 56),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 255), 2)
    cv2.putText(frame, f"Circles: {circle_count} | Rects: {rect_count}", (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX, 0.55, (180, 255, 180), 2)
    cv2.putText(frame, f"Latency: {elapsed_ms:.2f} ms ({current_fps:.0f} FPS)", (20, 104),
                cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 200, 100), 2)

    out_video.write(frame)

    # Save 3 sample frames (Original + Threshold + Detected) for visual inspection
    if frame_idx in [15, 60, 105]:
        sample_frames.append((frame_idx + 1, thresh.copy(), cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)))

    frame_idx += 1

cap.release()
out_video.release()

# ---------------------------------------------------------
# 3. Print Summary Metrics & Export Visualizations
# ---------------------------------------------------------
print("=== FRAME-BY-FRAME INFERENCE METRICS ===")
print(f"Total Frames Processed:      {total_frames}")
print(f"Average Latency per Frame:   {np.mean(inference_times_ms):.3f} ms")
print(f"Min / Max Latency:           {np.min(inference_times_ms):.3f} ms / {np.max(inference_times_ms):.3f} ms")
print(f"Average Throughput (FPS):    {np.mean(fps_list):.1f} FPS")
print(f"Mean Objects Detected/Frame: {np.mean(detected_counts):.2f} (Ground Truth: 4.00)")
print(f"Detection Accuracy Rate:     {(np.mean(detected_counts)/4.0)*100:.1f}%\n")

# Plot visual frames (Adaptive Threshold Mask vs. Final Output)
fig, axes = plt.subplots(len(sample_frames), 2, figsize=(10, 9))
for row_i, (f_num, mask_img, det_img) in enumerate(sample_frames):
    axes[row_i, 0].imshow(mask_img, cmap='gray')
    axes[row_i, 0].set_title(f"Frame {f_num}: Adaptive Threshold Mask")
    axes[row_i, 0].axis('off')

    axes[row_i, 1].imshow(det_img)
    axes[row_i, 1].set_title(f"Frame {f_num}: Contour Detection & Count Overlay")
    axes[row_i, 1].axis('off')

plt.tight_layout()
plt.savefig("pipeline_visual_frames.png", dpi=150)
plt.show()

# Plot Latency & FPS across all frames
fig, ax1 = plt.subplots(figsize=(9, 4))
ax1.plot(range(1, total_frames + 1), inference_times_ms, color='royalblue', label='Latency (ms)')
ax1.set_xlabel("Frame Number")
ax1.set_ylabel("Inference Latency (ms)", color='royalblue')
ax1.grid(True, linestyle='--', alpha=0.5)
plt.title("Frame-by-Frame Inference Latency Across Video Stream")
plt.tight_layout()
plt.savefig("performance_metrics.png", dpi=150)
plt.show()
```
