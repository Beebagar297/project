```python
# 1. Import the required tools
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score

# Download language dictionaries quietly
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

# 2. Expanded dataset (30 sentences with clear patterns)
texts = [
    # Positive (10)
    "I love this product, it works amazingly well!",
    "Best purchase I have made all year, very happy.",
    "The customer service was fantastic and super helpful.",
    "Absolutely thrilled with the fast delivery and great build!",
    "Great quality and amazing value for the money.",
    "So happy with this item, I love how easy it is to use.",
    "Fantastic experience, fast shipping and great support.",
    "This is wonderful, works great and feels very sturdy.",
    "Very happy with my order, amazing quality overall.",
    "Super helpful team and a really great product.",
    # Negative (10)
    "Terrible quality, it broke after just one day.",
    "I am very disappointed and want my money back.",
    "Worst experience ever, do not buy this.",
    "Awful software, keeps crashing every five minutes.",
    "Poor quality item, broke immediately and feels cheap.",
    "Very disappointed with the terrible customer service.",
    "Complete waste of money, worst product I have bought.",
    "It stopped working after two days, awful build quality.",
    "Do not waste your money, terrible and disappointing.",
    "Broken on arrival, worst delivery and poor support.",
    # Neutral (10)
    "The package arrived on Tuesday as scheduled.",
    "It is an average item, nothing special about it.",
    "The box contains one charger and a user manual.",
    "The product is okay, does what it says on the label.",
    "Standard delivery time, arrived in a brown box.",
    "It works as expected, neither good nor bad.",
    "The device comes in black and weighs two pounds.",
    "Average performance, it is an okay basic product.",
    "Received the order yesterday, standard packaging.",
    "The manual explains the basic setup steps clearly."
]

labels = (
    ["Positive"] * 10 +
    ["Negative"] * 10 +
    ["Neutral"] * 10
)

# 3. Clean the text (Remove stop-words + Lemmatization)
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(sentence):
    sentence = re.sub(r'[^a-zA-Z\s]', '', sentence.lower())
    words = sentence.split()
    cleaned_words = [lemmatizer.lemmatize(w) for w in words if w not in stop_words]
    return " ".join(cleaned_words)

cleaned_texts = [clean_text(t) for t in texts]

# 4. Convert words to numbers using TF-IDF
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(cleaned_texts)

# 5. Split data into training (70%) and testing (30%)
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.30, random_state=42, stratify=labels
)

# 6. Train the Scikit-Learn classifier
model = LogisticRegression()
model.fit(X_train, y_train)

# 7. Test the model and print the F1-score report
predictions = model.predict(X_test)
print("Overall Weighted F1-Score:", round(f1_score(y_test, predictions, average='weighted'), 2))
print("\nDetailed F1-Score Report:\n")
print(classification_report(y_test, predictions, zero_division=0))

# 8. Predict a brand-new sentence
new_sentence = ["I am really happy with how fast this runs!"]
new_cleaned = [clean_text(s) for s in new_sentence]
new_numbers = vectorizer.transform(new_cleaned)
print("Test Sentence:", new_sentence[0])
print("Predicted Sentiment:", model.predict(new_numbers)[0])
```
